# Mã hóa ký tự là gì?
____

Mã hóa ký tự là các bộ quy tắc cụ thể dùng để ánh xạ từ các chuỗi byte nhị phân thô (có dạng như thế này: 0110100001101001) sang các ký tự tạo nên văn bản mà con người có thể đọc được (ví dụ: "hi"). Có rất nhiều kiểu mã hóa khác nhau; nếu bạn cố gắng đọc một văn bản bằng kiểu mã hóa khác với kiểu được dùng để viết văn bản đó ban đầu, kết quả sẽ là những ký tự bị lỗi hiển thị, được gọi là "mojibake" (phát âm là *mo-gee-bah-kay*). Dưới đây là một ví dụ về mojibake:

æ–‡å—åŒ–ã??

Bạn cũng có thể gặp phải các ký tự "không xác định" (unknown characters). Đây là những gì hiển thị khi không có sự ánh xạ giữa một byte cụ thể và bất kỳ ký tự nào trong bảng mã mà bạn đang sử dụng để đọc chuỗi byte đó; chúng thường có dạng như sau:

����������

Ngày nay, tình trạng không tương thích về mã hóa ký tự đã ít phổ biến hơn trước, nhưng đây vẫn là một vấn đề cần lưu ý. Có rất nhiều kiểu mã hóa ký tự khác nhau, nhưng kiểu quan trọng nhất mà bạn cần biết là UTF-8.

> UTF-8 là chuẩn mã hóa văn bản phổ biến nhất. Mọi mã nguồn Python đều sử dụng UTF-8 và lý tưởng nhất là dữ liệu của bạn cũng nên được định dạng theo chuẩn này. Rắc rối thường nảy sinh khi dữ liệu không được mã hóa bằng UTF-8.

Việc xử lý mã hóa trong Python 2 từng khá phức tạp, nhưng rất may là trong Python 3, mọi thứ đã đơn giản hơn nhiều. (Kaggle Kernels chỉ sử dụng Python 3). Có hai kiểu dữ liệu chính mà bạn sẽ gặp khi làm việc với văn bản trong Python 3. Một trong số đó là kiểu chuỗi (string) – đây là định dạng mặc định cho văn bản.

```python
# start with a string
before = "This is the euro symbol: €"

# check to see what datatype it is
type(before)
```

Một loại dữ liệu khác là kiểu dữ liệu [bytes](https://docs.python.org/3.1/library/functions.html#bytes), vốn là một chuỗi các số nguyên. Bạn có thể chuyển đổi một chuỗi (string) sang dạng bytes bằng cách chỉ định bảng mã (encoding) của nó:

```python
# encode it to a different encoding, replacing characters that raise errors
after = before.encode("utf-8", errors = "replace")

# check the type
type(after)
```

Khi quan sát một đối tượng kiểu `bytes`, bạn sẽ thấy nó có ký tự `b` ở phía trước, theo sau là một đoạn văn bản. Lý do là vì dữ liệu kiểu `bytes` được hiển thị giống như các ký tự được mã hóa theo chuẩn ASCII (ASCII là một chuẩn mã hóa ký tự cũ, không thực sự phù hợp để thể hiện các ngôn ngữ khác ngoài tiếng Anh). Ở đây, bạn có thể thấy ký hiệu đồng euro đã bị thay thế bằng một chuỗi ký tự lỗi (mojibake) trông giống như `\xe2\x82\xac` khi được hiển thị dưới dạng chuỗi ASCII.

```python
# take a look at what the bytes look like
after
```

Khi chuyển đổi lại các byte thành chuỗi văn bản với đúng bảng mã, chúng ta thấy toàn bộ nội dung văn bản đều hiển thị chính xác – thật tuyệt! :)

```python
# convert it back to utf-8
print(after.decode("utf-8"))
```

Tuy nhiên, khi cố gắng sử dụng một bảng mã (encoding) khác để chuyển đổi chuỗi byte thành chuỗi ký tự (string), ta sẽ gặp lỗi. Nguyên nhân là do bảng mã đó không biết cách xử lý các byte mà ta đang truyền vào. Bạn cần chỉ định cho Python biết bảng mã thực sự tương ứng với chuỗi byte đó.

> Bạn có thể hình dung các bảng mã khác nhau giống như những phương thức ghi âm khác nhau. Cùng một bản nhạc có thể được ghi lại trên đĩa CD, băng cassette hoặc băng 8-track. Dù âm thanh nghe có vẻ gần như nhau, bạn vẫn cần sử dụng thiết bị phù hợp để phát nhạc từ từng định dạng ghi âm đó. Bộ giải mã (decoder) chính xác đóng vai trò như máy phát băng cassette hay đầu đĩa CD vậy; nếu bạn cố phát băng cassette trên đầu đĩa CD, nó sẽ không thể hoạt động được.

```python
# try to decode our bytes with the ascii encoding
print(after.decode("ascii"))
```

Chúng ta cũng có thể gặp rắc rối nếu cố gắng sử dụng sai bảng mã để chuyển đổi từ chuỗi (string) sang dạng byte. Như đã đề cập trước đó, trong Python 3, các chuỗi mặc định sử dụng mã hóa UTF-8; do đó, nếu ta cố xử lý chúng như thể chúng đang dùng một bảng mã khác, vấn đề sẽ nảy sinh.

Ví dụ, nếu ta dùng phương thức `encode()` để chuyển đổi một chuỗi sang dạng byte theo chuẩn ASCII, ta đang yêu cầu hệ thống tạo ra chuỗi byte tương ứng với văn bản đó trong bảng mã ASCII. Tuy nhiên, vì văn bản gốc không phải là ASCII, sẽ có những ký tự mà bảng mã này không thể xử lý được. Ta có thể chọn giải pháp tự động thay thế các ký tự không được ASCII hỗ trợ. Tuy nhiên, khi làm vậy, mọi ký tự nằm ngoài phạm vi ASCII sẽ bị thay thế bằng ký tự đại diện cho giá trị không xác định. Sau đó, khi chuyển đổi ngược lại từ byte sang chuỗi, ký tự ban đầu sẽ bị thay thế bằng ký tự không xác định đó. Điều nguy hiểm ở đây là không có cách nào để biết ký tự gốc *đáng lẽ* phải là gì. Điều này đồng nghĩa với việc dữ liệu của chúng ta có thể đã trở nên vô dụng!

```python
# start with a string
before = "This is the euro symbol: €"

# encode it to a different encoding, replacing characters that raise errors
after = before.encode("ascii", errors = "replace")

# convert it back to utf-8
print(after.decode("ascii"))

# Chúng ta đã mất chuỗi byte gốc ban đầu! Nó đã bị
# thay thế bằng chuỗi byte dành cho ký tự không xác định :(
```

Đây là điều không tốt và chúng ta cần tránh làm vậy! Tốt hơn nhiều là chuyển đổi toàn bộ văn bản sang UTF-8 càng sớm càng tốt và duy trì định dạng mã hóa đó. Thời điểm lý tưởng nhất để chuyển đổi dữ liệu đầu vào (không phải UTF-8) sang UTF-8 là ngay khi đọc tệp tin; chúng ta sẽ thảo luận về vấn đề này ở phần tiếp theo.

Tuy nhiên, trước tiên, hãy thử chuyển đổi qua lại giữa kiểu byte và chuỗi (string) với các bảng mã khác nhau để xem kết quả ra sao. Hãy quan sát những thay đổi xảy ra với văn bản của bạn. Liệu bạn có muốn điều này xảy ra với dữ liệu mà mình đang định phân tích hay không?

```python
# Your turn! Try encoding and decoding different symbols to ASCII and
# see what happens. I'd recommend $, #, 你好 and नमस्ते but feel free to
# try other characters. What happens? When would this cause problems?
```

```python
# Danh sách các ký tự/chuỗi cần thử nghiệm
test_symbols = ['$', '#', '你好', 'नमस्ते', 'café', '🚀']

print("=== THỬ NGHIỆM MÃ HÓA (ENCODE) & GIẢI MÃ (DECODE) ASCII ===\n")

for symbol in test_symbols:
    print(f"Ký tự ban đầu: '{symbol}'")
    
    # 1. Thử mã hóa trực tiếp sang ASCII (Mặc định strict mode)
    try:
        encoded_strict = symbol.encode('ascii')
        print(f"  └─ Mã hóa strict: {encoded_strict}")
        decoded_strict = encoded_strict.decode('ascii')
        print(f"  └─ Giải mã strict: '{decoded_strict}'")
    except UnicodeEncodeError as e:
        print(f"  └─ Mã hóa strict: LỖI (UnicodeEncodeError) -> Không thể mã hóa bằng ASCII!")

    # 2. Xử lý ký tự ngoài ASCII bằng các chế độ đệm (error handlers)
    encoded_ignore = symbol.encode('ascii', errors='ignore')
    encoded_replace = symbol.encode('ascii', errors='replace')
    encoded_escaped = symbol.encode('ascii', errors='backslashreplace')

    print(f"  └─ Xử lý 'ignore'           : {encoded_ignore} -> Giải mã: '{encoded_ignore.decode('ascii')}'")
    print(f"  └─ Xử lý 'replace'          : {encoded_replace} -> Giải mã: '{encoded_replace.decode('ascii')}'")
    print(f"  └─ Xử lý 'backslashreplace' : {encoded_escaped} -> Giải mã: '{encoded_escaped.decode('ascii')}'")
    print("-" * 60)
```

# Đọc các tệp gặp vấn đề về mã hóa
___

Hầu hết các tệp bạn gặp sẽ thường được mã hóa bằng UTF-8. Đây là định dạng mặc định mà Python sử dụng, vì vậy phần lớn thời gian bạn sẽ không gặp vấn đề gì. Tuy nhiên, đôi khi bạn có thể gặp lỗi như sau:

```python
# try to read in a file not in UTF-8
kickstarter_2016 = pd.read_csv("../input/kickstarter-projects/ks-projects-201612.csv")
```

Hãy để ý rằng chúng ta gặp lại lỗi `UnicodeDecodeError` y hệt như khi cố gắng giải mã các byte UTF-8 theo chuẩn ASCII! Điều này cho thấy tệp tin thực chất không được mã hóa bằng UTF-8. Tuy nhiên, chúng ta lại không biết chính xác nó sử dụng kiểu mã hóa nào. Một cách để xác định là thử nghiệm lần lượt nhiều kiểu mã hóa ký tự khác nhau xem có cái nào hoạt động hay không. Dù vậy, cách tốt hơn là sử dụng mô-đun `chardet` để tự động đoán kiểu mã hóa phù hợp. Phương pháp này không đảm bảo chính xác 100%, nhưng thường nhanh hơn nhiều so với việc tự đoán mò.

Tôi sẽ chỉ xem xét 10.000 byte đầu tiên của tệp tin này. Thông thường, lượng dữ liệu này là đủ để đưa ra dự đoán chính xác về kiểu mã hóa và nhanh hơn nhiều so với việc kiểm tra toàn bộ tệp tin (đặc biệt là với các tệp tin lớn, việc kiểm tra toàn bộ có thể rất chậm). Một lý do khác để chỉ xem phần đầu tệp tin là dựa vào thông báo lỗi, ta thấy vấn đề đầu tiên nằm ở ký tự thứ 11. Do đó, có lẽ chúng ta chỉ cần xem xét một đoạn ngắn ở đầu tệp tin là đủ để hiểu chuyện gì đang xảy ra.

```python
# look at the first ten thousand bytes to guess the character encoding
with open("../input/kickstarter-projects/ks-projects-201801.csv", 'rb') as rawdata:
    result = chardet.detect(rawdata.read(10000))

# check what the character encoding might be
print(result)
```

So chardet is 73%  confidence that the right encoding is "Windows-1252". Let's see if that's correct:

```python
# read in the file with the encoding detected by chardet
kickstarter_2016 = pd.read_csv("../input/kickstarter-projects/ks-projects-201612.csv", encoding='Windows-1252')

# look at the first few lines
kickstarter_2016.head()
```

Đúng vậy, có vẻ như `chardet` đã đoán đúng! Việc đọc tệp diễn ra suôn sẻ (dù có xuất hiện cảnh báo về kiểu dữ liệu) và khi xem xét vài dòng đầu tiên, mọi thứ có vẻ đều ổn.

> **Nếu mã hóa mà `chardet` dự đoán không chính xác thì sao?** Vì về cơ bản `chardet` chỉ là một công cụ dự đoán thông minh, nên đôi khi nó có thể đoán sai mã hóa. Một cách bạn có thể thử là thay đổi phạm vi dữ liệu đọc từ tệp (xem nhiều hơn hoặc ít hơn) để kiểm tra xem có nhận được kết quả khác hay không, rồi thử áp dụng kết quả đó.
